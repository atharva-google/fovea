## 4. Upload a video

```python
from google.colab import files
uploaded = files.upload()
VIDEO = list(uploaded.keys())[0]
VIDEO
```

## 5. Reframe
`targets` accepts ratios (`"9:16"`, `"1:1"`, `"4:5"`), exact sizes (`"1080x1920"`) or presets (`"vertical"`, `"square"`, `"portrait"`).
Analysis runs once and is reused for every target.

```python
from fovea import run

outputs = run(VIDEO, targets=["9:16", "1:1"])
outputs
```

Equivalent CLI: `!python -m fovea run my_video.mp4 -t 9:16 1:1`

## 6. Preview & download

```python
from fovea.utils.notebook import show_video
show_video(outputs[0])
```

```python
for o in outputs:
    files.download(str(o))
```
